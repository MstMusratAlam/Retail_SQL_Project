```python
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas


owner_name = "Mr Yusuf"
phone_number = "+8801575251404"

advance_amount = 7500
monthly_rent = 75000


month = input("Enter Month (Example: April 2026): ")

file_name = f"Rent_Memo_{month.replace(' ','_')}.pdf"


c = canvas.Canvas(file_name, pagesize=letter)

c.setFont("Helvetica-Bold", 16)
c.drawString(180, 750, "RENT PAYMENT RECEIPT")

c.setFont("Helvetica", 12)

c.drawString(100, 700, f"Owner Name: {owner_name}")
c.drawString(100, 675, f"Phone: {phone_number}")

c.drawString(100, 640, f"Month: {month}")

c.drawString(100, 600, f"Advance (1 Month): BDT {advance_amount}")
c.drawString(100, 575, f"Monthly Rent: BDT {monthly_rent}")

total = advance_amount + monthly_rent
c.drawString(100, 540, f"Total Paid: BDT {total}")

c.drawString(100, 500, "Status: Paid")

c.drawString(100, 450, "Generated by Rental Memo System")

c.save()

print("PDF Created Successfully:", file_name)
```
